```python
%matplotlib widget
import gc
from helpers import NV_from_settings, settings, filename_from_settings
n = 512
```

# Lid driven Cavity

```python
settings_lid = settings()
settings_lid.update({"addon": "lid", "nx":n, "ny":n,
                      "T_max":30, "type_":"lid", "solver":"GPU2"})
NV_test = NV_from_settings(settings_lid)
NV_test.save(filename_from_settings(settings_lid, subfolder="presentation"))
del NV_test
gc.collect()
```
